```python
import os
import json
import re
from collections import Counter

```

```python
# Input from Mileston
EXTRACTED_DIR = "/content/extracted_sections"


# Output for Milestone 4
OUTPUT_DIR = "/content/Milestone4/summaries"
os.makedirs(OUTPUT_DIR, exist_ok=True)

PAPER_SUMMARY_FILE = os.path.join(OUTPUT_DIR, "paper_summaries.json")
FINAL_REVIEW_FILE = os.path.join(OUTPUT_DIR, "final_review.txt")
```

```python
def clean_text(text):
    text = re.sub(r"\s+", " ", text)
    text = re.sub(r"[^a-zA-Z0-9., ]", "", text)
    return text.strip()
```

```python
def summarize_text(text, max_sentences=3):
    sentences = text.split(".")
    sentences = [s.strip() for s in sentences if len(s.strip()) > 30]

    if len(sentences) <= max_sentences:
        return ". ".join(sentences)

    words = re.findall(r"\w+", text.lower())
    word_freq = Counter(words)

    sentence_scores = {}
    for sent in sentences:
        for word in sent.lower().split():
            if word in word_freq:
                sentence_scores[sent] = sentence_scores.get(sent, 0) + word_freq[word]

    ranked = sorted(sentence_scores, key=sentence_scores.get, reverse=True)
    return ". ".join(ranked[:max_sentences])
```

```python
papers_data = {}

for file in os.listdir(EXTRACTED_DIR):
    if file.endswith(".json"):
        with open(os.path.join(EXTRACTED_DIR, file), "r", encoding="utf-8") as f:
            papers_data[file.replace(".json", "")] = json.load(f)

print("Loaded papers:", list(papers_data.keys()))
```
